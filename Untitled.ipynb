```python
"""
PIPELINE ETL DATA ENGINEERING - ITALIANO PERFETTO
Funziona con QUALSIASI dataset CSV
"""

import pandas as pd
import sqlite3
import os
import matplotlib.pyplot as plt
from datetime import datetime

print("🚀 === PIPELINE ETL DATA ENGINEERING ITALIANA === 🚀")
print(f"Avvio: {datetime.now().strftime('%d/%m %H:%M')}")

# 1. ESTRAZIONE
print("\n📥 PASSO 1: Caricamento CSV")
df = pd.read_csv('healthcare_dataset.csv')
print(f"✅ {len(df):,} righe, {len(df.columns)} colonne")
print("Colonne:", list(df.columns))

os.makedirs('bronze', exist_ok=True)
os.makedirs('silver', exist_ok=True)

# 2. BRONZE
print("\n🏅 PASSO 2: Livello Bronze")
df.to_csv('bronze/dati_grezzi.csv', index=False)

# 3. SILVER
print("\n🔧 PASSO 3: Livello Silver")
df_silver = df.copy()

numeriche = df.select_dtypes(include='number').columns
categoriche = df.select_dtypes(include='object').columns

print(f"Numeriche: {list(numeriche)}")
print(f"Categoriche: {list(categoriche)}")

if len(numeriche) > 0:
    col_num = numeriche[0]
    df_silver['valore_alto'] = df_silver[col_num] > df_silver[col_num].median()

df_silver.dropna(inplace=True)
df_silver.to_csv('silver/dati_puliti.csv', index=False)

# 4. GRAFICI (COLORI STANDARD)
print("\n📈 PASSO 4: Grafici")
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
fig.suptitle('Dashboard ETL Italiana')

# Grafico 1
if len(categoriche) > 0:
    top = df[categoriche[0]].value_counts().head(8)
    top.plot(kind='bar', ax=axes[0,0], color='red')  # ROSSO
    axes[0,0].set_title(f'Top {categoriche[0]}')
    axes[0,0].tick_params(axis='x', rotation=45)

# Grafico 2
if len(numeriche) > 0:
    df[numeriche[0]].hist(ax=axes[0,1], color='blue', bins=20)  # BLU
    axes[0,1].set_title(numeriche[0])

# Grafico 3
if len(categoriche) > 0 and len(numeriche) > 0:
    media = df.groupby(categoriche[0])[numeriche[0]].mean().head(8)
    media.plot(kind='bar', ax=axes[1,0], color='green')  # VERDE
    axes[1,0].set_title(f'Media {numeriche[0]}')
    axes[1,0].tick_params(axis='x', rotation=45)

# Grafico 4
if 'valore_alto' in df_silver.columns:
    df_silver['valore_alto'].value_counts().plot(kind='pie', ax=axes[1,1], autopct='%1.1f%%')
    axes[1,1].set_title('Valori Alti %')

plt.tight_layout()
plt.savefig('dashboard_italiana.png', dpi=300)
plt.show()

# 5. WAREHOUSE
print("\n🏢 PASSO 5: Data Warehouse")
conn = sqlite3.connect('warehouse.db')
df_silver.to_sql('dati_clean', conn, if_exists='replace')

if len(categoriche) > 0 and len(numeriche) > 0:
    query = f"SELECT {categoriche[0]}, COUNT(*), AVG({numeriche[0]}) FROM dati_clean GROUP BY {categoriche[0]} LIMIT 10"
    risultati = pd.read_sql(query, conn)
    risultati.to_csv('risultati.csv', index=False)
    print(risultati)

conn.close()

print("\n🎉 PIPELINE TERMINATA!")
print("File:")
print("- bronze/dati_grezzi.csv")
print("- silver/dati_puliti.csv")
print("- warehouse.db")
print("- dashboard_italiana.png")
print("- risultati.csv")
```
